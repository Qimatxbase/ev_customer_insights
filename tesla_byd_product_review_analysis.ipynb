```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Data
# -----------------------------
brands = ["BYD", "Tesla"]

h1_2025 = np.array([8556, 14146])
h1_2026 = np.array([29192, 23588])

# Increase from H1 2025 to H1 2026
increase = h1_2026 - h1_2025

# YoY growth reported by CarExpert
growth_rate = [241.2, 66.7]

# -----------------------------
# Create chart
# -----------------------------
fig, ax = plt.subplots(figsize=(8, 6))

x = np.arange(len(brands))
width = 0.55

# H1 2025 baseline
bars_2025 = ax.bar(
    x,
    h1_2025,
    width,
    label="H1 2025 deliveries"
)

# Additional deliveries in H1 2026
bars_growth = ax.bar(
    x,
    increase,
    width,
    bottom=h1_2025,
    label="Increase in H1 2026"
)

# -----------------------------
# Labels inside bars
# -----------------------------
for i in range(len(brands)):

    # H1 2025 value
    ax.text(
        x[i],
        h1_2025[i] / 2,
        f"H1 2025\n{h1_2025[i]:,}",
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold"
    )

    # Increase value
    ax.text(
        x[i],
        h1_2025[i] + increase[i] / 2,
        f"+{increase[i]:,}",
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold"
    )

    # H1 2026 total + YoY growth
    ax.text(
        x[i],
        h1_2026[i] + 800,
        f"{h1_2026[i]:,}\n(+{growth_rate[i]:.1f}% YoY)",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold"
    )

# -----------------------------
# Formatting
# -----------------------------
ax.set_title(
    "Growth in Australian EV Deliveries: BYD vs Tesla",
    fontsize=15,
    fontweight="bold",
    pad=18
)

ax.set_ylabel("EV Deliveries")
ax.set_xticks(x)
ax.set_xticklabels(brands, fontsize=11)

ax.set_ylim(0, 35000)

ax.grid(
    axis="y",
    alpha=0.25
)

ax.set_axisbelow(True)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    frameon=False,
    loc="upper right"
)

# -----------------------------
# Source
# -----------------------------
fig.text(
    0.5,
    0.01,
    "Source: Author-created figure based on CarExpert (2025, 2026), "
    "using FCAI and Electric Vehicle Council data.",
    ha="center",
    fontsize=8
)

plt.tight_layout(rect=[0, 0.05, 1, 1])

plt.savefig(
    "BYD_Tesla_EV_Growth_H1_2025_2026.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
